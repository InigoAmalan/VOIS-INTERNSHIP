```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from IPython.display import display

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

print("=" * 70)
print("SEASONAL AGRICULTURE PERFORMANCE ANALYSIS")
print("=" * 70)



print("\nDATASET INFORMATION")
print("-" * 70)

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

print("\nFirst 5 rows:")
display(df.head())



print("\nDATA CLEANING")
print("-" * 70)

duplicate_count = df.duplicated().sum()

print("Duplicate rows found:", duplicate_count)

if duplicate_count > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")


missing = df.isnull().sum()

print("\nMissing values:")
display(missing[missing > 0])

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].median())


categorical_columns = df.select_dtypes(include='object').columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("\nMissing values after cleaning:")
print(df.isnull().sum().sum())




print("\nBASIC STATISTICS")
print("-" * 70)

display(df.describe())

print("\nCategorical Variables:")
for column in categorical_columns:
    print(f"\n{column}")
    print(df[column].value_counts())



if 'Profit_INR' in df.columns and 'Total_Cost_INR' in df.columns:

    df['ROI_Percentage'] = np.where(
        df['Total_Cost_INR'] != 0,
        (df['Profit_INR'] / df['Total_Cost_INR']) * 100,
        0
    )

print("\nROI calculated successfully.")



print("\n" + "=" * 70)
print("SEASONAL AGRICULTURAL PERFORMANCE")
print("=" * 70)

season_summary = df.groupby('Season').agg(
    Records=('Season', 'count'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Cost=('Total_Cost_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean'),
    Average_Water_Used=('Water_Used_m3', 'mean'),
    Average_Water_Efficiency=('Water_Efficiency_t_per_1000m3', 'mean'),
    Average_Disease_Risk=('Disease_Pest_Risk_pct', 'mean'),
    Average_ROI=('ROI_Percentage', 'mean')
).round(2)

display(season_summary)




plt.figure(figsize=(9,6))

sns.countplot(
    data=df,
    x='Season',
    order=df['Season'].value_counts().index
)

plt.title("Number of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Yield_Tonnes_Ha',
    estimator='mean'
)

plt.title("Average Crop Yield Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Profit_INR',
    estimator='mean'
)

plt.title("Average Profit Across Seasons")
plt.xlabel("Season")
plt.ylabel("Profit (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Revenue_INR',
    estimator='mean'
)

plt.title("Average Revenue Across Seasons")
plt.xlabel("Season")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Total_Cost_INR',
    estimator='mean'
)

plt.title("Average Agricultural Cost Across Seasons")
plt.xlabel("Season")
plt.ylabel("Cost (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



print("\n" + "=" * 70)
print("RESOURCE USAGE ANALYSIS")
print("=" * 70)


plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Water_Used_m3',
    estimator='mean'
)

plt.title("Average Water Usage Across Seasons")
plt.xlabel("Season")
plt.ylabel("Water Used (m³)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Water_Efficiency_t_per_1000m3',
    estimator='mean'
)

plt.title("Water Efficiency Across Seasons")
plt.xlabel("Season")
plt.ylabel("Tonnes per 1000 m³")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Fertilizer_kg_ha',
    estimator='mean'
)

plt.title("Average Fertilizer Usage Across Seasons")
plt.xlabel("Season")
plt.ylabel("Fertilizer (kg/ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Pesticide_Litre_ha',
    estimator='mean'
)

plt.title("Average Pesticide Usage Across Seasons")
plt.xlabel("Season")
plt.ylabel("Pesticide (Litre/ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()




print("\n" + "=" * 70)
print("ENVIRONMENTAL CONDITIONS")
print("=" * 70)



plt.figure(figsize=(9,6))

sns.boxplot(
    data=df,
    x='Season',
    y='Rainfall_mm'
)

plt.title("Rainfall Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Rainfall (mm)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.boxplot(
    data=df,
    x='Season',
    y='Avg_Temperature_C'
)

plt.title("Temperature Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.boxplot(
    data=df,
    x='Season',
    y='Humidity_pct'
)

plt.title("Humidity Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Humidity (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.boxplot(
    data=df,
    x='Season',
    y='Soil_Moisture_pct'
)

plt.title("Soil Moisture Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Soil Moisture (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()




plt.figure(figsize=(9,6))

sns.barplot(
    data=df,
    x='Season',
    y='Disease_Pest_Risk_pct',
    estimator='mean'
)

plt.title("Average Disease and Pest Risk by Season")
plt.xlabel("Season")
plt.ylabel("Risk (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



print("\n" + "=" * 70)
print("CROP PERFORMANCE ANALYSIS")
print("=" * 70)

crop_summary = df.groupby('Crop').agg(
    Records=('Crop', 'count'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean'),
    Average_ROI=('ROI_Percentage', 'mean')
).sort_values(
    'Average_Yield',
    ascending=False
).round(2)

display(crop_summary)


plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x='Crop',
    order=df['Crop'].value_counts().index
)

plt.title("Agricultural Records by Crop")
plt.xlabel("Crop")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x='Crop',
    y='Yield_Tonnes_Ha',
    estimator='mean',
    order=df.groupby('Crop')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False).index
)

plt.title("Average Yield by Crop")
plt.xlabel("Crop")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



crop_season_yield = pd.pivot_table(
    df,
    values='Yield_Tonnes_Ha',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

print("\nAverage Yield by Crop and Season:")
display(crop_season_yield.round(2))


plt.figure(figsize=(12,8))

sns.heatmap(
    crop_season_yield,
    annot=True,
    fmt='.2f'
)

plt.title("Average Crop Yield by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()




crop_season_profit = pd.pivot_table(
    df,
    values='Profit_INR',
    index='Crop',
    columns='Season',
    aggfunc='mean'
)

print("\nAverage Profit by Crop and Season:")
display(crop_season_profit.round(2))


plt.figure(figsize=(12,8))

sns.heatmap(
    crop_season_profit,
    annot=True,
    fmt='.0f'
)

plt.title("Average Crop Profit by Crop and Season")
plt.xlabel("Season")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()



print("\n" + "=" * 70)
print("REGIONAL PERFORMANCE")
print("=" * 70)

state_summary = df.groupby('State').agg(
    Records=('State', 'count'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Production=('Production_Tonnes', 'mean'),
    Average_Revenue=('Revenue_INR', 'mean'),
    Average_Profit=('Profit_INR', 'mean')
).sort_values(
    'Average_Profit',
    ascending=False
).round(2)

display(state_summary)




print("\n" + "=" * 70)
print("IRRIGATION ANALYSIS")
print("=" * 70)

irrigation_summary = df.groupby('Irrigation_Method').agg(
    Records=('Irrigation_Method', 'count'),
    Average_Yield=('Yield_Tonnes_Ha', 'mean'),
    Average_Water_Used=('Water_Used_m3', 'mean'),
    Average_Profit=('Profit_INR', 'mean')
).sort_values(
    'Average_Yield',
    ascending=False
).round(2)

display(irrigation_summary)


plt.figure(figsize=(11,6))

sns.barplot(
    data=df,
    x='Irrigation_Method',
    y='Yield_Tonnes_Ha',
    estimator='mean'
)

plt.title("Average Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()



print("\n" + "=" * 70)
print("RELATIONSHIP ANALYSIS")
print("=" * 70)



plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='Rainfall_mm',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title("Rainfall vs Crop Yield")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='Avg_Temperature_C',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title("Temperature vs Crop Yield")
plt.xlabel("Temperature (°C)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()



plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='Soil_Moisture_pct',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title("Soil Moisture vs Crop Yield")
plt.xlabel("Soil Moisture (%)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()


plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='Fertilizer_kg_ha',
    y='Yield_Tonnes_Ha',
    hue='Season'
)

plt.title("Fertilizer Usage vs Crop Yield")
plt.xlabel("Fertilizer (kg/ha)")
plt.ylabel("Yield (Tonnes/Ha)")
plt.tight_layout()
plt.show()




print("\n" + "=" * 70)
print("CORRELATION ANALYSIS")
print("=" * 70)

numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(18,14))

sns.heatmap(
    correlation_matrix,
    cmap='coolwarm',
    center=0
)

plt.title("Correlation Matrix of Agricultural Variables")
plt.tight_layout()
plt.show()



yield_corr = correlation_matrix[
    'Yield_Tonnes_Ha'
].sort_values(
    ascending=False
)

print("\nVariables correlated with Yield:")
display(yield_corr.round(3))


profit_corr = correlation_matrix[
    'Profit_INR'
].sort_values(
    ascending=False
)

print("\nVariables correlated with Profit:")
display(profit_corr.round(3))



print("\n" + "=" * 70)
print("STATISTICAL SIGNIFICANCE TEST")
print("=" * 70)

seasons = df['Season'].dropna().unique()

yield_groups = [
    df[df['Season'] == season]['Yield_Tonnes_Ha'].dropna()
    for season in seasons
]

if len(yield_groups) >= 2:

    yield_anova = stats.f_oneway(*yield_groups)

    print("ANOVA Statistic :", round(yield_anova.statistic, 4))
    print("P-value         :", round(yield_anova.pvalue, 6))

    if yield_anova.pvalue < 0.05:
        print("\nRESULT:")
        print("There is a statistically significant difference")
        print("in crop yield across seasons.")
    else:
        print("\nRESULT:")
        print("There is no statistically significant difference")
        print("in crop yield across seasons.")




profit_groups = [
    df[df['Season'] == season]['Profit_INR'].dropna()
    for season in seasons
]

if len(profit_groups) >= 2:

    profit_anova = stats.f_oneway(*profit_groups)

    print("\nProfit ANOVA Statistic :", round(profit_anova.statistic, 4))
    print("Profit P-value         :", round(profit_anova.pvalue, 6))

    if profit_anova.pvalue < 0.05:
        print("\nRESULT:")
        print("There is a statistically significant difference")
        print("in agricultural profit across seasons.")
    else:
        print("\nRESULT:")
        print("There is no statistically significant difference")
        print("in agricultural profit across seasons.")



print("\n" + "=" * 70)
print("OUTLIER DETECTION")
print("=" * 70)

Q1 = df['Yield_Tonnes_Ha'].quantile(0.25)
Q3 = df['Yield_Tonnes_Ha'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

yield_outliers = df[
    (df['Yield_Tonnes_Ha'] < lower_limit) |
    (df['Yield_Tonnes_Ha'] > upper_limit)
]

print("Lower limit :", round(lower_limit, 2))
print("Upper limit :", round(upper_limit, 2))
print("Number of yield outliers :", len(yield_outliers))




plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x='Yield_Tonnes_Ha',
    kde=True
)

plt.title("Distribution of Crop Yield")
plt.xlabel("Yield (Tonnes/Ha)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()




plt.figure(figsize=(10,6))

sns.boxplot(
    data=df,
    x='Season',
    y='Yield_Tonnes_Ha'
)

plt.title("Yield Distribution Across Seasons")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()




print("\n" + "=" * 70)
print("KEY FINDINGS")
print("=" * 70)

best_yield_season = season_summary['Average_Yield'].idxmax()
worst_yield_season = season_summary['Average_Yield'].idxmin()

best_profit_season = season_summary['Average_Profit'].idxmax()
worst_profit_season = season_summary['Average_Profit'].idxmin()

best_revenue_season = season_summary['Average_Revenue'].idxmax()

best_water_efficiency_season = (
    season_summary['Average_Water_Efficiency'].idxmax()
)

highest_risk_season = (
    season_summary['Average_Disease_Risk'].idxmax()
)

best_roi_season = season_summary['Average_ROI'].idxmax()


print("\n1. Highest average yield:")
print("   ", best_yield_season)

print("\n2. Lowest average yield:")
print("   ", worst_yield_season)

print("\n3. Highest average profit:")
print("   ", best_profit_season)

print("\n4. Lowest average profit:")
print("   ", worst_profit_season)

print("\n5. Highest average revenue:")
print("   ", best_revenue_season)

print("\n6. Best water efficiency:")
print("   ", best_water_efficiency_season)

print("\n7. Highest disease/pest risk:")
print("   ", highest_risk_season)

print("\n8. Highest ROI:")
print("   ", best_roi_season)



best_crop_yield = crop_summary['Average_Yield'].idxmax()
best_crop_profit = crop_summary['Average_Profit'].idxmax()
best_crop_roi = crop_summary['Average_ROI'].idxmax()

print("\nBEST CROPS")

print("\nHighest yielding crop:")
print("   ", best_crop_yield)

print("\nHighest profit crop:")
print("   ", best_crop_profit)

print("\nHighest ROI crop:")
print("   ", best_crop_roi)




best_state_profit = state_summary['Average_Profit'].idxmax()
best_state_yield = state_summary['Average_Yield'].idxmax()

print("\nBEST REGIONS")

print("\nHighest average profit:")
print("   ", best_state_profit)

print("\nHighest average yield:")
print("   ", best_state_yield)




print("\n" + "=" * 70)
print("FINAL SEASONAL SUMMARY")
print("=" * 70)

final_summary = season_summary.copy()

display(final_summary)



print("\n" + "=" * 70)
print("DATA-DRIVEN RECOMMENDATIONS")
print("=" * 70)

print(f"""
1. Seasonal Planning
   - The season with the highest average yield is {best_yield_season}.
   - Agricultural planning can consider the performance observed
     during this season.

2. Economic Planning
   - The season with the highest average profit is {best_profit_season}.
   - Farmers and planners can examine the cost and revenue conditions
     associated with this season.

3. Water Management
   - {best_water_efficiency_season} shows the highest average water
     efficiency.
   - Water management practices used during this period can be studied
     for efficient resource utilization.

4. Disease and Pest Management
   - {highest_risk_season} has the highest average disease/pest risk.
   - Additional monitoring and preventive agricultural practices may
     be considered during this period.

5. Crop Selection
   - {best_crop_yield} has the highest average yield among the crops
     present in the dataset.
   - Its seasonal performance can be examined when planning crop
     selection.

6. Regional Planning
   - {best_state_profit} has the highest average profit among the
     regions/states represented in the dataset.
   - Its agricultural conditions and practices can be investigated
     further.

7. Evidence-Based Decision Making
   - Seasonal agricultural decisions should consider yield, profit,
     resource usage, environmental conditions and disease/pest risk
     together rather than relying on a single metric.
""")




fig, axes = plt.subplots(2, 2, figsize=(16,10))


sns.barplot(
    data=df,
    x='Season',
    y='Yield_Tonnes_Ha',
    estimator='mean',
    ax=axes[0,0]
)

axes[0,0].set_title("Average Yield by Season")
axes[0,0].set_xlabel("Season")
axes[0,0].set_ylabel("Yield (Tonnes/Ha)")


sns.barplot(
    data=df,
    x='Season',
    y='Profit_INR',
    estimator='mean',
    ax=axes[0,1]
)

axes[0,1].set_title("Average Profit by Season")
axes[0,1].set_xlabel("Season")
axes[0,1].set_ylabel("Profit (INR)")



sns.barplot(
    data=df,
    x='Season',
    y='Water_Used_m3',
    estimator='mean',
    ax=axes[1,0]
)

axes[1,0].set_title("Average Water Usage by Season")
axes[1,0].set_xlabel("Season")
axes[1,0].set_ylabel("Water Used (m³)")



sns.barplot(
    data=df,
    x='Season',
    y='Disease_Pest_Risk_pct',
    estimator='mean',
    ax=axes[1,1]
)

axes[1,1].set_title("Average Disease/Pest Risk by Season")
axes[1,1].set_xlabel("Season")
axes[1,1].set_ylabel("Risk (%)")


plt.suptitle(
    "Seasonal Agriculture Performance Dashboard",
    fontsize=18
)

plt.tight_layout()
plt.show()



season_summary.to_csv(
    'seasonal_performance_summary.csv'
)

crop_summary.to_csv(
    'crop_performance_summary.csv'
)

state_summary.to_csv(
    'state_performance_summary.csv'
)

correlation_matrix.to_csv(
    'agriculture_correlation_matrix.csv'
)

print("\n" + "=" * 70)
print("ANALYSIS COMPLETED SUCCESSFULLY!")
print("=" * 70)

print("""
Files generated:

1. seasonal_performance_summary.csv
2. crop_performance_summary.csv
3. state_performance_summary.csv
4. agriculture_correlation_matrix.csv
""")
```
